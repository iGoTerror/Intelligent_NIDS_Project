# Intelligent Web-Based Network Intrusion Detection System Using Machine Learning

## Notebook 2: Data Preprocessing

### Objective

The objective of this notebook is to transform the raw NSL-KDD dataset into a machine learning-ready dataset.

This notebook includes:

- Loading the dataset
- Removing unnecessary features
- Converting attack labels
- Separating features and target
- Encoding categorical variables
- Scaling numerical features
- Preparing the final dataset for Support Vector Machine (SVM)

At the end of this notebook, the dataset will be ready for model training.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import os
```

```python
PROJECT_PATH = "/content/drive/MyDrive/Intelligent_NIDS_Project"

DATASET_PATH = PROJECT_PATH + "/datasets"
MODEL_PATH = PROJECT_PATH + "/models"
RESULTS_PATH = PROJECT_PATH + "/results"
```

```python
column_names = [
    'duration','protocol_type','service','flag','src_bytes','dst_bytes',
    'land','wrong_fragment','urgent','hot','num_failed_logins',
    'logged_in','num_compromised','root_shell','su_attempted',
    'num_root','num_file_creations','num_shells','num_access_files',
    'num_outbound_cmds','is_host_login','is_guest_login','count',
    'srv_count','serror_rate','srv_serror_rate','rerror_rate',
    'srv_rerror_rate','same_srv_rate','diff_srv_rate',
    'srv_diff_host_rate','dst_host_count','dst_host_srv_count',
    'dst_host_same_srv_rate','dst_host_diff_srv_rate',
    'dst_host_same_src_port_rate','dst_host_srv_diff_host_rate',
    'dst_host_serror_rate','dst_host_srv_serror_rate',
    'dst_host_rerror_rate','dst_host_srv_rerror_rate',
    'label','difficulty'
]

train_df = pd.read_csv(
    DATASET_PATH + "/KDDTrain+.txt",
    names=column_names
)

test_df = pd.read_csv(
    DATASET_PATH + "/KDDTest+.txt",
    names=column_names
)

print("Training Dataset Shape:", train_df.shape)
print("Testing Dataset Shape:", test_df.shape)

train_df.head()
```

```python
train_df.info()
```

```python
train_df.isnull().sum()
```

```python
train_df.duplicated().sum()
```

```python
X_train = train_df.drop(columns=["label", "difficulty"])
y_train = train_df["label"]

X_test = test_df.drop(columns=["label", "difficulty"])
y_test = test_df["label"]
```

```python
print(X_train.shape)
print(y_train.shape)
```

```python
y_train = y_train.apply(lambda x: 0 if x == "normal" else 1)
y_test = y_test.apply(lambda x: 0 if x == "normal" else 1)
```

```python
print(y_train.value_counts())
```

```python
print("Protocol Types:")
print(train_df["protocol_type"].unique())

print("\nServices:")
print(train_df["service"].unique())

print("\nFlags:")
print(train_df["flag"].unique())
```

```python
categorical_columns = ["protocol_type", "service", "flag"]

X_train = pd.get_dummies(
    X_train,
    columns=categorical_columns
)

X_test = pd.get_dummies(
    X_test,
    columns=categorical_columns
)
```

```python
print("Training Features Shape:", X_train.shape)
print("Testing Features Shape:", X_test.shape)
```

```python
X_train.head()
```

```python
X_train, X_test = X_train.align(
    X_test,
    join="left",
    axis=1,
    fill_value=0
)
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
```

```python
X_train_scaled = scaler.fit_transform(X_train)
```

```python
X_test_scaled = scaler.transform(X_test)
```

```python
print(X_train_scaled.shape)
print(X_test_scaled.shape)
```
