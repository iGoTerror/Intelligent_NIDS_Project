# Notebook 07: Systematic SVM Experimental Investigation & Optimization

## Executive Overview
This notebook presents the systematic empirical evaluation of **Support Vector Machines (SVM)** for Network Intrusion Detection (NIDS) using the official `KDDTest+` dataset ($N = 22,544$).

### Experimental Scope (Experiments A through F):
1. **Experiment A**: Kernel Selection (Linear vs Poly vs RBF vs Sigmoid)
2. **Experiment B**: Penalty Regularization Parameter ($C \in [0.1, 1000.0]$)
3. **Experiment C**: Kernel Gamma Parameter ($\gamma \in [\text{scale}, 0.001, 0.01, 0.1, 1.0]$)
4. **Experiment D**: Class Weight Balancing Strategies (`None`, `balanced`, `custom_1:2`, `custom_1:3`)
5. **Experiment E**: Feature Subset Dimensionality ($k \in [10, 20, 30, 40, 50, 75, 100, 122]$)
6. **Experiment F**: Decision Boundary Probability Threshold Calibration ($P \in [0.01, 0.99]$)


```python
import os
import json
import numpy as np
import pandas as pd

RESULTS_PATH = r"../results"
MODEL_PATH = r"../models"
```

## 1. Load & Display Experimental Result Matrices (Experiments A - F)


```python
exp_a = pd.read_csv(os.path.join(RESULTS_PATH, "experiment_a_kernels.csv"))
exp_b = pd.read_csv(os.path.join(RESULTS_PATH, "experiment_b_c_values.csv"))
exp_c = pd.read_csv(os.path.join(RESULTS_PATH, "experiment_c_gamma_values.csv"))
exp_d = pd.read_csv(os.path.join(RESULTS_PATH, "experiment_d_class_weighting.csv"))
exp_e = pd.read_csv(os.path.join(RESULTS_PATH, "experiment_e_feature_counts.csv"))
exp_f = pd.read_csv(os.path.join(RESULTS_PATH, "experiment_f_threshold_sweep.csv"))

print("=== EXPERIMENT A: KERNEL SELECTION ===")
display(exp_a[["Kernel", "Attack_Recall_%", "Attack_Precision_%", "Attack_F1_%", "FPR_%", "Accuracy_%"]])

print("\n=== EXPERIMENT B: REGULARIZATION PARAMETER C ===")
display(exp_b[["C", "Attack_Recall_%", "Attack_Precision_%", "Attack_F1_%", "FPR_%", "Accuracy_%"]])

print("\n=== EXPERIMENT E: FEATURE SUBSET SIZE k ===")
display(exp_e[["k_features", "Attack_Recall_%", "Attack_Precision_%", "Attack_F1_%", "FPR_%", "Accuracy_%"]])
```
